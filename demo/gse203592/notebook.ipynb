Why: Core measure: each guide's S13/S07 share and own-gene change vs control

```python

import numpy as np, pandas as pd, scipy.sparse as sp
from scipy.stats import spearmanr, pearsonr
a = pt.screen().adata
obs = a.obs.copy()
cs = pt.load('20261113T002746-C-states_and_composition-9f25d3:cell_states').set_index('barcode')
obs['state'] = cs.reindex(obs.index)['state'].values
obs['guide'] = obs['guides'].astype(str).str.split(':').str[0]
lab = obs['label_after_qc'].astype(str)
targets = ['ARID1A','SMARCC1','SMARCD2','ARID1B','ARID2','ACTR5','INO80C','GATA3','PDCD1']
# lognorm of raw counts
X = a.X
tot = np.asarray(X.sum(axis=1)).ravel()
D = sp.diags(1e4/np.maximum(tot,1))
L = (D @ X).tocsr(); L.data = np.log1p(L.data)
gcol = {g.upper(): i for i,g in enumerate(a.var_names)}
def gene_vec(sym):
    i = gcol[sym.upper()]
    return np.asarray(L[:, i].todense()).ravel()
own = {t: gene_vec(t) for t in targets}
ctrl = obs[lab=='control']
ctrl_s13 = (ctrl.state=='S13').mean(); ctrl_s07=(ctrl.state=='S07').mean()
print('control S13', ctrl_s13, 'S07', ctrl_s07, 'n', len(ctrl))
rows=[]; unit_rows=[]
ctrl_unit = ctrl.groupby('unit').apply(lambda d: pd.Series({'s13':(d.state=='S13').mean(),'s07':(d.state=='S07').mean(),'n':len(d)}))
for t in targets:
    m = (lab==t)&(obs.guide.str.startswith(t+'-'))
    for g in sorted(obs.loc[m,'guide'].unique()):
        mg = m & (obs.guide==g)
        d = obs[mg]
        ow = own[t][mg.values]; oc = own[t][(lab=='control').values]
        r = dict(target=t, guide=g, n=len(d), s13=(d.state=='S13').mean(), s07=(d.state=='S07').mean(),
                 d_s13=(d.state=='S13').mean()-ctrl_s13, d_s07=(d.state=='S07').mean()-ctrl_s07,
                 own_lfc=ow.mean()-oc.mean(), frac_own_zero=(ow==0).mean())
        # per unit
        ups13=[];ups07=[]
        for u in sorted(d.unit.unique()):
            du = d[d.unit==u]
            if len(du)<10: continue
            cu = ctrl[ctrl.unit==u]
            ds13=(du.state=='S13').mean()-(cu.state=='S13').mean()
            ds07=(du.state=='S07').mean()-(cu.state=='S07').mean()
            unit_rows.append(dict(target=t,guide=g,unit=u,n=len(du),d_s13=ds13,d_s07=ds07))
        rows.append(r)
G = pd.DataFrame(rows)
U = pd.DataFrame(unit_rows)
pd.set_option('display.width',250)
print(G.round(4).to_string())
```

**Per-guide Mantel-Haenszel test of S13 and S07 against control cells, by unit** (01:39, t12-swi-snf-shift), on thread t12: r5 (SWI/SNF loss depletes S13 and S10, raises S07): does the shared SWI/SNF response hold per guide and per unit, and does ARID1B/ARID2 (paralog, weaker) show a dose-ordered version of it by own-gene knockdown? Test: guides view on ARID1A/SMARCC1/SMARCD2/ARID1B/ARID2, S13/S07 share and own-gene change per guide against control-guide groups.

Why: Per-guide significance for S13 and S07 with units as strata

```python

import numpy as np, pandas as pd
from statsmodels.stats.contingency_tables import StratifiedTable
a = pt.screen().adata
obs = a.obs.copy()
cs = pt.load('20261113T002746-C-states_and_composition-9f25d3:cell_states').set_index('barcode')
obs['state'] = cs.reindex(obs.index)['state'].values
obs['guide'] = obs['guides'].astype(str).str.split(':').str[0]
lab = obs['label_after_qc'].astype(str)
ctrl = obs[lab=='control']
out=[]
for g in sorted([x for x in obs.guide.unique() if x.split('-')[0] in ['ARID1A','SMARCC1','ARID1B','ARID2','SMARCD2'] and x.split('-')[0]==x.split('-')[0]]):
    t = g.split('-')[0]
    gd = obs[(obs.guide==g)&(lab==t)]
    for st_ in ['S13','S07']:
        tabs=[]; 
        for u in sorted(gd.unit.unique()):
            gu = gd[gd.unit==u]; cu = ctrl[ctrl.unit==u]
            if len(gu)<10 or len(cu)<10: continue
            a1=(gu.state==st_).sum(); b1=len(gu)-a1
            a0=(cu.state==st_).sum(); b0=len(cu)-a0
            tabs.append(np.array([[a1,b1],[a0,b0]])+0.5)
        if len(tabs)<2: continue
        T = StratifiedTable(tabs)
        r = T.test_null_odds(); ci = T.oddsratio_pooled_confint()
        out.append(dict(guide=g, state=st_, units=len(tabs), OR=round(T.oddsratio_pooled,3), ci=(round(ci[0],3),round(ci[1],3)), p=float('%.3g'%r.pvalue)))
R = pd.DataFrame(out)
print(R.to_string())
pt.save_table('t12__per_guide_MH', R)
```

```python
import numpy as np, pandas as pd
a = pt.screen().adata
st = pt.load("20261113T002746-C-states_and_composition-9f25d3:cell_states").set_index('barcode')
obs = a.obs.copy()
obs['state'] = st['state'].reindex(obs.index.astype(str)).values if False else None
bc = obs.index.astype(str)
m = st['state'].reindex(bc)
print('matched', m.notna().sum(), 'of', len(bc))
print(obs['label_after_qc'].value_counts().to_dict())
g = list(a.var_names).index('Pdcd1')
L = pt.screen().lognorm
x = np.asarray(L[:, g].todense()).ravel() if hasattr(L,'todense') else np.asarray(L[:, g]).ravel()
df = pd.DataFrame({'label': obs['label_after_qc'].values, 'unit': obs['unit'].values, 'state': m.values, 'pdcd1': x})
df = df[df['label'].isin(['control','GATA3','PDCD1','ARID1A']) & df['state'].notna()]
rows=[]
ctrl = df[df.label=='control']
wC = ctrl.groupby('state').size()/len(ctrl)
muC = ctrl.groupby('state')['pdcd1'].mean()
for t in ['GATA3','PDCD1','ARID1A']:
    d = df[df.label==t]
    wT = d.groupby('state').size()/len(d)
    muT = d.groupby('state')['pdcd1'].mean()
    states = sorted(set(wT.index)|set(wC.index))
    wT=wT.reindex(states).fillna(0); wC=wC.reindex(states).fillna(0)
    muT=muT.reindex(states); muC=muC.reindex(states)
    within_s = (muT-muC)
    within = float((wT*within_s.fillna(0)).sum())
    comp = float(((wT-wC)*muC.fillna(0)).sum())
    total = float(d.pdcd1.mean()-ctrl.pdcd1.mean())
    rows.append({'target':t,'total_diff':total,'within_state':within,'composition':comp,'sum_check':within+comp})
    for s in ['S05','S06','S10','S01','S04']:
        rows.append({'target':t,'state':s,'share_target':float(wT.get(s,0)),'share_ctrl':float(wC.get(s,0)),'mu_target':float(muT.get(s,np.nan)),'mu_ctrl':float(muC.get(s,np.nan)),'within_diff':float(within_s.get(s,np.nan)),'n_target':int((d.state==s).sum())})
out = pd.DataFrame(rows)
pd.set_option('display.width',250)
print(out.round(4).to_string())
pt.save_table("t13__decomposition", out)
```

**S13 share difference per unit using single-guide cells only versus all cells** (02:08)

```python
import pandas as pd, numpy as np
o = pt.screen().adata.obs.copy()
cs = pt.load('20261113T012359-C-states_and_composition-1b6d86:cell_states').set_index('barcode')
o['state']=cs.state.reindex(o.index)
o = o[o.label_after_qc.notna() & (o.label_after_qc!='') & o.state.notna()]
o['multi']=o.guides.str.contains(';',na=False)
print(o.label_after_qc.value_counts().to_dict()); print('multi frac',o.multi.mean().round(3))
rows=[]
ctl = [l for l in o.label_after_qc.unique() if 'CTRL' in str(l).upper() or 'non' in str(l).lower()]
print('controls',ctl)
for tgt in ['ARID1A','ARID1B','SMARCD2','ACTR5','PDCD1']:
    for subset,name in [(o,'all'),(o[~o.multi],'single')]:
        d=[]
        for u,g in subset.groupby('unit'):
            t=g[g.label_after_qc==tgt]; c=g[g.label_after_qc.isin(ctl)]
            if len(t)>=10 and len(c)>=10:
                d.append((t.state.eq('S13').mean()-c.state.eq('S13').mean(), len(t)))
        w=np.array([x[1] for x in d]); v=np.array([x[0] for x in d])
        rows.append(dict(target=tgt,cells=name,units=len(d),neg=int((v<0).sum()),mean_diff=v.mean(),wmean=(v*w).sum()/w.sum(),n_cells=int(w.sum())))
r=pd.DataFrame(rows).round(4); print(r.to_string()); pt.save_table('t15__s13_single_guide',r)
```
